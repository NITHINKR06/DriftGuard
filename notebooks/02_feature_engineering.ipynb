# Phase 2: Feature Engineering
Input: processed_phase1.csv
Goal: Create ML-ready behavioral features


```python
import pandas as pd

df = pd.read_csv("../data/processed_phase1.csv")
print(df.shape)
df.head()
```

Separate Labels & Features

```python
labels = df['binary_label']
df_features = df.drop(columns=[' Label', 'binary_label'])
```

Remove Non-Numeric Columns

```python
df_features = df_features.select_dtypes(include=['int64', 'float64'])
print(df_features.shape)
```

🔑 Isolation Forest & Autoencoders require numeric-only input

Feature Scaling (CRITICAL)

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_features)
```

```python
import numpy as np

np.save("../data/X_phase2.npy", X_scaled)
labels.to_numpy().dump("../data/y_phase2.npy")

print("Phase 2 features saved")
```

## Phase 2 Summary (Saved)

- Removed non-numeric features
- Standardized feature distributions
- Prepared ML-ready feature matrix
- Feature pipeline modularized
- Saved scaled features for model training
